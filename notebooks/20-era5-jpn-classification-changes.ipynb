# ERA5 country WBGT classification changes
Configure a cached-country analysis without using the CLI.

```python
from datetime import date
from pathlib import Path

from IPython.display import display
from liljegren_analysis.classification import analyze_era5_country_cache, write_outputs

REPOSITORY = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
COUNTRY_CODE = 'JPN'
START_DATE = date(2025, 1, 1)
END_DATE = date(2025, 12, 31)
THRESHOLDS_C = (21, 25, 28, 31)
CACHE_PATH = REPOSITORY / 'data/country' / COUNTRY_CODE.lower() / 'observations.parquet'
OUTPUT_ROOT = REPOSITORY / 'data/country_classification' / COUNTRY_CODE.lower()
```

```python
ANALYSIS = analyze_era5_country_cache(CACHE_PATH, START_DATE, END_DATE, THRESHOLDS_C)
write_outputs(ANALYSIS.outputs, OUTPUT_ROOT)
display(ANALYSIS.observation_summary)
display(ANALYSIS.quantity_coverage)
display(ANALYSIS.classification.classes)
display(ANALYSIS.classification.outputs['pair_change_summary.csv'])
display(ANALYSIS.classification.outputs['monthly_pair_change_summary.csv'])
```

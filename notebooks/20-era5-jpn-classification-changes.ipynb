# ERA5 country WBGT classification changes

This notebook analyzes an existing country observation cache. Country/cell selection is deliberately separate: create or refresh the cache with `scripts/extract_country.py` first.

```python
from datetime import date
from pathlib import Path

from IPython.display import display

from liljegren_analysis.classification import analyze_era5_country_cache, write_outputs
```

```python
# COUNTRY_CODE identifies a pre-extracted cache only.
COUNTRY_CODE = 'JPN'
ANALYSIS_START_DATE = date(2025, 1, 1)
ANALYSIS_END_DATE = date(2025, 12, 31)
WBGT_THRESHOLDS_C = (21, 25, 28, 31)  # JPN MoE exercise/activity bands
```

```python
repository = next(
    (candidate for candidate in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
     if (candidate / 'pyproject.toml').is_file()),
    None,
)
if repository is None:
    raise FileNotFoundError('Could not locate the repository root')
country_code = COUNTRY_CODE.strip().lower()
cache_path = repository / 'data' / 'country' / country_code / 'observations.parquet'
output_directory = repository / 'data' / 'country_classification' / country_code
analysis = analyze_era5_country_cache(
    cache_path, ANALYSIS_START_DATE, ANALYSIS_END_DATE, WBGT_THRESHOLDS_C
)
display(analysis.observation_summary)
display(analysis.quantity_coverage)
display(analysis.classification.classes)
display(analysis.classification.outputs['pair_change_summary.csv'])
display(analysis.classification.outputs['monthly_pair_change_summary.csv'])
```

```python
write_outputs(analysis.outputs, output_directory)
print(f'Saved {len(analysis.outputs)} classification outputs to {output_directory}')
```

# NASA POWER WBGT classification changes

The reusable module loads the workload, filters locations/countries, and compares matched implementation results.

```python
from datetime import date
from pathlib import Path

from IPython.display import display

from liljegren_analysis.classification import analyze_nasa_power_run, write_outputs
```

```python
RUN_DIRECTORY = Path('.hsb/runs/wbgt-comparison-20260827-144607')
SHARD_ORDINAL = 1
CYCLE_ORDINAL = 1
ANALYSIS_START_DATE = date(2001, 1, 1)
ANALYSIS_END_DATE = date(2025, 12, 31)
COUNTRY_CODES = ['SG']  # Set to None to retain all locations.
LOCATION_IDS = ['singapore']  # Set to None to retain all locations.
WBGT_THRESHOLDS_C = (31, 32, 33)
```

```python
repository = next(
    (candidate for candidate in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
     if (candidate / 'pyproject.toml').is_file()),
    None,
)
if repository is None:
    raise FileNotFoundError('Could not locate the repository root')
run_directory = RUN_DIRECTORY if RUN_DIRECTORY.is_absolute() else repository / RUN_DIRECTORY
country_scope = '-'.join(code.lower() for code in COUNTRY_CODES) if COUNTRY_CODES else 'all'
output_directory = repository / 'data' / 'country_classification' / country_scope
analysis = analyze_nasa_power_run(
    run_directory, ANALYSIS_START_DATE, ANALYSIS_END_DATE, WBGT_THRESHOLDS_C,
    shard_ordinal=SHARD_ORDINAL, cycle_ordinal=CYCLE_ORDINAL,
    country_codes=COUNTRY_CODES, location_ids=LOCATION_IDS,
    boundary_cache=repository / 'data' / 'boundaries',
)
display(analysis.workload_summary)
display(analysis.status_summary)
display(analysis.classification.classes)
display(analysis.classification.outputs['pair_change_summary.csv'])
display(analysis.classification.outputs['monthly_pair_change_summary.csv'])
```

```python
write_outputs(analysis.outputs, output_directory)
print(f'Saved {len(analysis.outputs)} classification outputs to {output_directory}')
```

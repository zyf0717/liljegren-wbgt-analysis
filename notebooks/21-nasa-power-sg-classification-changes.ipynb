# NASA POWER WBGT classification changes
Configure a run analysis without using the CLI.

```python
from datetime import date
from pathlib import Path

from IPython.display import display
from liljegren_analysis.classification import analyze_nasa_power_run, write_outputs

REPOSITORY = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
RUN_ROOT = REPOSITORY / '.hsb/runs/wbgt-comparison-20260827-144607'
OUTPUT_ROOT = REPOSITORY / 'data/country_classification/sg'
BOUNDARY_CACHE = REPOSITORY / 'data/boundaries'
START_DATE = date(2001, 1, 1)
END_DATE = date(2025, 12, 31)
THRESHOLDS_C = (31, 32, 33)
SHARD_ORDINAL = 1
CYCLE_ORDINAL = 1
COUNTRY_CODES = ('SG',)
LOCATION_IDS = ('singapore',)
```

```python
ANALYSIS = analyze_nasa_power_run(
    RUN_ROOT, START_DATE, END_DATE, THRESHOLDS_C,
    shard_ordinal=SHARD_ORDINAL, cycle_ordinal=CYCLE_ORDINAL,
    country_codes=COUNTRY_CODES, location_ids=LOCATION_IDS,
    boundary_cache=BOUNDARY_CACHE,
)
write_outputs(ANALYSIS.outputs, OUTPUT_ROOT)
display(ANALYSIS.workload_summary)
display(ANALYSIS.status_summary)
display(ANALYSIS.classification.classes)
display(ANALYSIS.classification.outputs['pair_change_summary.csv'])
display(ANALYSIS.classification.outputs['monthly_pair_change_summary.csv'])
```

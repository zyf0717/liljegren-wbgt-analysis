# Joint-support cellwise temperature quantiles

Generate the persisted cellwise WBGT, Tg, and Tnwb quantile product. The notebook configures and invokes the public pipeline API; it does not contain analysis logic.

```python
from datetime import date
from pathlib import Path

from liljegren_analysis import (
    DEFAULT_DATE_WORKERS, DEFAULT_JOINT_IMPLEMENTATIONS, DEFAULT_JOINT_OUTPUT_ROOT,
    DEFAULT_REPLAY_WORKERS, compute_joint_cellwise_temperature_quantiles,
)

repository = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
HSB_RUNS = [repository.parent / 'HeatStressBench/.hsb/runs/wbgt-comparison-20260902-150355']
OUTPUT_ROOT = repository / DEFAULT_JOINT_OUTPUT_ROOT
START_DATE = date(2025, 1, 1)
END_DATE = date(2025, 12, 31)
IMPLEMENTATIONS = DEFAULT_JOINT_IMPLEMENTATIONS
DATE_WORKERS = DEFAULT_DATE_WORKERS
REPLAY_WORKERS = DEFAULT_REPLAY_WORKERS
REPLACE_EXISTING_OUTPUT = False
```

```python
output = compute_joint_cellwise_temperature_quantiles(
    HSB_RUNS,
    start_date=START_DATE,
    end_date=END_DATE,
    output_root=OUTPUT_ROOT,
    implementations=IMPLEMENTATIONS,
    date_workers=DATE_WORKERS,
    replay_workers=REPLAY_WORKERS,
    overwrite=REPLACE_EXISTING_OUTPUT,
)
output
```
